### ASESMEN NON KOGNITIF GAYA BELAJAR SISWA MTsN 1 KOTA PADANG TAHUN PELAJAR 2024-2025

```python
import pandas as pd
```

```python
df=pd.read_excel("data_belajar.xlsx")
df.head()
```

```python
df=df.drop(index=0).reset_index(drop=True)
```

```python
print(df.columns.tolist())
```

```python
df=df.drop(columns=['Timestamp', 'Score', 'Nama Lengkap : ', 'Alamat : ', 'Asal Sekolah : '])
```

```python
df.head()
```

```python
df.insert(0, "No", range(1, len(df) + 1))
```

```python
df.head()
```

```python
df.columns = ['No'] + [str(i) for i in range(1, len(df.columns))]
for col in df.columns[1:]:  # skip kolom 'No'
    df[col] = df[col].astype(str).str.strip().str.upper().str[0]
```

```python
df.head()
```

```python
df_jenis = df.copy()
df_jenis["Visual"] = (df_jenis.iloc[:, 1:] == "A").sum(axis=1)
df_jenis["Auditori"] = (df_jenis.iloc[:, 1:] == "B").sum(axis=1)
df_jenis["Kinestetik"] = (df_jenis.iloc[:, 1:] == "C").sum(axis=1)
```

```python
def dominan(row):
    jenis = {"Visual": row["Visual"], "Auditori": row["Auditori"], "Kinestetik": row["Kinestetik"]}
    return max(jenis, key=jenis.get)

df_jenis["Dominan"] = df_jenis.apply(dominan, axis=1)
```

```python
gaya_counts = df_jenis["Dominan"].value_counts()
print(gaya_counts)
```

```python
import matplotlib.pyplot as plt

# Pie chart
plt.figure(figsize=(6, 6))
plt.pie(gaya_counts, labels=gaya_counts.index, autopct='%1.1f%%', startangle=140)
plt.title("Distribusi Gaya Belajar Siswa")
plt.show()

# Bar chart
plt.figure(figsize=(8, 4))
gaya_counts.plot(kind='bar', color=['skyblue', 'salmon', 'lightgreen'])
plt.title("Jumlah Siswa per Gaya Belajar")
plt.xlabel("Gaya Belajar")
plt.ylabel("Jumlah Siswa")
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
df_jenis.to_excel("hasil_analisis_gaya_belajar.xlsx", index=False)
```
